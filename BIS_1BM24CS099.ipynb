```python
# Network Path Optimization using Genetic Algorithm

# Step 1: Initialization
paths = [
    ("S -> A -> B -> D", 18),
    ("S -> C -> D", 11),
    ("S -> A -> C -> D", 14),
    ("S -> B -> D", 20)
]

print("Network Path Optimization Using GA\n")

print("Step 1: Possible Paths")
for path, energy in paths:
    print(path, "Energy:", energy)

# Step 2: Fitness Evaluation
print("\nStep 2: Fitness Evaluation")
for path, energy in paths:
    fitness = 1 / (1 + energy)
    print(path, "Fitness:", round(fitness, 4))

# Step 3: Selection
best_path = min(paths, key=lambda x: x[1])

print("\nStep 3: Selection")
print("Selected Path:", best_path[0])

# Step 4: Optimal Path
print("\nStep 4: Optimal Path")
print("Source: S")
print("Receiver: D")
print("Optimal Path:", best_path[0])
print("Minimum Energy:", best_path[1], "units")
```
